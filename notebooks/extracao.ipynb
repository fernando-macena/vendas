```python
import requests #biblioteca responsável por fazer uma requisição a API
import json #biblioteca responsável por salvar o arquivo
from requests import Response #classe importada para facilitar a implementacao da função salvar_resposta (poderia utilizar somente requests.Response)
```

```python
url_base = "https://transparencia.api.ro.gov.br/api/v1/remuneracao-servidor"
```

```python
#A funcao obter dados é a responsável por montar a URL com os parametros e fazer a requisicao.
#Ela retorna um objeto do tipo Response.

def obter_dados(ano: int, mes: int, sigla_ug: str, cargo: str, page: int = 1, page_size: int = 10):
  parametros = {
      "Ano": ano,
      "Mes": mes,
      "SiglaUg": sigla_ug,
      "Cargo": cargo,
      "Page": page,
      "PageSize": page_size
  }
  return requests.get(url_base, params=parametros)
```

```python
#A funcao obter parametros de pesquisa é a responsável por obter alguns parametros de uma forma interativa
#com o usuário, permitindo que ele, o usuário do serviço, possa customizar suas
#consultas.

def obter_parametros_de_pesquisa():
  ano = input("Digite o ano: ") or 2023,
  mes = input("Digite o mês: ") or 1,
  sigla_ug = input("Digite a sigla do orgao publico que deseja: ") or "CGE",
  cargo = input("Digite o cargo: ") or "AUDITOR DE CONTROLE INTERNO"

  return ano, mes, sigla_ug, cargo
```

```python
def salvar_resposta(resposta: Response):
  with open("dados_raw.json", "w") as arquivo:
    json.dump(resposta.json(), arquivo)
```

```python
#Funcao desativada para facilitar o desenvolvimento do ETL
#obter_resposta = obter_dados(*obter_parametros_de_pesquisa())

#Por enquando os parametros serao passados de forma harcoded
obter_resposta = obter_dados(2023, 1, "CGE", "AUDITOR DE CONTROLE INTERNO")

salvar_resposta(obter_resposta)

print(type(obter_resposta))
```
